```python
df2=pd.read_csv("customer_id_list.csv")
```

```python
df
```

```python
clients = pd.merge(df, df2, on=['FIRST_NAME',"LAST_NAME","GENDER","AGE"])
```

```python
df3=pd.read_csv("purchase_log.csv")
```

```python
df3
```

```python
clients
```

```python
detailed_sales=pd.merge(df3,clients,on=["CUSTOMER_ID"],how="inner")
```

```python
detailed_sales
```

```python
spenders=detailed_sales.groupby(["FIRST_NAME", "LAST_NAME"])["PRICE"].sum().reset_index()
```

```python
spenders_sorted = spenders.sort_values(by="PRICE", ascending=False)
```

```python
spenders_sorted.head(1)
```

```python
top_spender_first = spenders_sorted.iloc[0]['FIRST_NAME']
top_spender_last = spenders_sorted.iloc[0]['LAST_NAME']
```

```python
top_client_purchases = detailed_sales[(detailed_sales['FIRST_NAME'] == top_spender_first) & (detailed_sales['LAST_NAME'] == top_spender_last)]
```

```python
top_client_purchases['ITEM_DESCRIPTION'].unique()
```

### Writing Files

```python
spenders_sorted.to_csv("top_spenders.csv", index=False)
```
